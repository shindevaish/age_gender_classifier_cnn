### Removing all images greather than 90 from the dataset

```python
import os
```

```python
dataset = os.listdir('dataset')
len(dataset)
```

```python
count_gt_90 = 0
for img_name in dataset:
    parts = img_name.split('_')
    age = int(parts[0])

    if age > 90:
        try:
            os.remove(os.path.join('dataset', img_name))
            count_gt_90 += 1    
        except Exception as e:
            print(f"Exception occured:{e}")
```
