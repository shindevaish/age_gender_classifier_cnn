```python
from PIL import Image, ImageFilter
import os
import time
import numpy as np
```

```python
dir = "dataset"

dataset = os.listdir(dir)
len(dataset)
```

```python
for img_name in dataset:
    parts = img_name.split("_")
    gender = int(parts[1])
    if gender > 1:
        print(img_name)
```

```python
output_dir = "augmented_dataset_v2"
os.makedirs(output_dir, exist_ok=True)
```

```python
def augment_and_save(image_path):
    # os.makedirs(output_dir, exist_ok=True)

    img = Image.open(image_path).convert("RGB")

    # Base name - [age]_[gender]_[race].jpg
    filename = os.path.basename(image_path)
    base_name = filename.split(".")[0]

    rotations = {
        "20_c": -20,   # clockwise
        "40_c": -40,
        "20_ac": 20,   # anticlockwise
        "40_ac": 40
    }

    # Saving original image
    img.save(os.path.join(output_dir, f"{base_name}.jpg"))

    for suffix, angle in rotations.items():
        rotated = img.rotate(angle, expand=False)
        rotated.save(os.path.join(output_dir, f"{base_name}_{suffix}.jpg"))

    # Rotating and saving mirrored image
    mirrored = img.transpose(Image.FLIP_LEFT_RIGHT)
    mirrored.save(os.path.join(output_dir, f"{base_name}_m.jpg"))

    for suffix, angle in rotations.items():
        rotated_mirror = mirrored.rotate(angle, expand=False)
        rotated_mirror.save(
            os.path.join(output_dir, f"{base_name}_m_{suffix}.jpg")
        )

    # #Slightly Blurred image
    # blurred = img.filter(ImageFilter.GaussianBlur(radius=1))
    # blurred.save(os.path.join(output_dir, f"{base_name}_blur.jpg"))

    # #Pixel Noise
    # img_np = np.array(img)gi
    # noise = np.random.normal(0, 8, img_np.shape).astype(np.int16)
    # noisy_img = img_np.astype(np.int16) + noise
    # noisy_img = np.clip(noisy_img, 0, 255).astype(np.uint8)
    
    # noisy_img_pil = Image.fromarray(noisy_img)
    # noisy_img_pil.save(os.path.join(output_dir, f"{base_name}_noise.jpg"))
```

```python
count = 0
start_time = time.time()

for img_path in dataset:
    count += 1
    augment_and_save(os.path.join(dir, img_path))

end_time = time.time()

print(f"Time taken for augmentation: {end_time - start_time} seconds")
print(f"Total images processed: {count}")
print("Data augmentation completed.")
```
