```python
from torchvision import transforms
from PIL import Image
import os

from data_loader import FaceDataset
```

```python
dir = "dataset"

dataset = os.listdir(dir)
len(dataset)
```

```python
output_dir = "augmented_dataset"
os.makedirs(output_dir, exist_ok=True)
```

```python
def augment_and_save(image_path):
    # os.makedirs(output_dir, exist_ok=True)

    img = Image.open(image_path).convert("RGB")

    # Base name - [age]_[gender]_[race].jpg
    filename = os.path.basename(image_path)
    base_name = "_".join(filename.split("_")[:3])

    rotations = {
        "20_c": -20,   # clockwise
        "40_c": -40,
        "20_ac": 20,   # anticlockwise
        "40_ac": 40
    }

    # Saving original image
    img.save(os.path.join(output_dir, f"{base_name}.jpg"))

    for suffix, angle in rotations.items():
        rotated = img.rotate(angle, expand=False)
        rotated.save(os.path.join(output_dir, f"{base_name}_{suffix}.jpg"))

    # Rotating and saving mirrored image
    mirrored = img.transpose(Image.FLIP_LEFT_RIGHT)
    mirrored.save(os.path.join(output_dir, f"{base_name}_m.jpg"))

    for suffix, angle in rotations.items():
        rotated_mirror = mirrored.rotate(angle, expand=False)
        rotated_mirror.save(
            os.path.join(output_dir, f"{base_name}_m_{suffix}.jpg")
        )
```

```python
for img_path in dataset:
    augment_and_save(os.path.join(dir, img_path))
print("Data augmentation completed.")
```
